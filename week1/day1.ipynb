## Types of prompts

You may know this already - but if not, you will get very familiar with it!

Models like GPT have been trained to receive instructions in a particular way.

They expect to receive:

**A system prompt** that tells them what task they are performing and what tone they should use

**A user prompt** -- the conversation starter that they should reply to

````python
# Define our system prompt - you can experiment with this later, changing the last sentence to 'Respond in markdown in Spanish."

system_prompt = """
You are a snarky assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
````

````python
# Define our user prompt

user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""
````

## Messages

The API from OpenAI expects to receive messages in a particular structure.
Many of the other APIs share this structure:

```python
[
    {"role": "system", "content": "system message goes here"},
    {"role": "user", "content": "user message goes here"}
]
```
To give you a preview, the next 2 cells make a rather simple call - we won't stretch the mighty GPT (yet!)

````python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "What is 2 + 2?"},
]

response = openai.chat.completions.create(model=openai_model, messages=messages)
response.choices[0].message.content
````

## And now let's build useful messages for GPT-4.1-mini, using a function

````python
# See how this function creates exactly the format above

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]
````

````python
# Try this out, and then try for a few more websites

messages_for(ed)
````

## Time to bring it together - the API for OpenAI is very simple!

````python
# And now: call the OpenAI API. You will get very familiar with this!
def summarize(url):
    website = fetch_website_contents(url)
    response = openai.chat.completions.create(
        model=openai_model, messages=messages_for(website)
    )
    return response.choices[0].message.content
````

````python
summarize("https://edwarddonner.com")
````

````python
# A function to display this nicely in the output, using markdown

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
````

````python
display_summary("https://edwarddonner.com")
````

````python
# Step 1: Create your prompts

system_prompt = """

You are a patient coding-interview mentor working with a junior developer.

Your goal: the student can solve a *similar, unseen* problem alone next week.
A correct solution in this transcript is not success. Every section must move
the student toward being able to re-derive the answer without you.

## Source of truth

- Work only from the problem statement the student pastes. If they give only a
  URL, ask for the full text (title, description, examples, constraints) instead
  of reconstructing it from memory.
- If a detail is missing or ambiguous, say so and state the assumption you are
  proceeding under.
- Never invent LeetCode metadata: no acceptance rates, no "beats X% of
  submissions," no claimed benchmark timings.

## Method: derivation over delivery

1. Restate the problem in plain language before doing anything else.
2. Start from the approach a beginner would actually reach for first — usually
   brute force. Establish that it is *correct* before criticizing its speed.
3. Improve only in response to a specific, named inefficiency. Every
   optimization must answer three questions explicitly:
   - What exact work were we repeating or throwing away?
   - What do we store or reorder to stop repeating it?
   - What does that cost us? (usually memory)
4. Introduce a data structure at the moment it solves the problem you just
   named — never earlier. In one or two sentences, say what it is and which
   single operation makes it the right tool (e.g. "a hash map gives average
   O(1) lookup by key").
5. Never present the optimal solution as something you simply knew. The student
   must see the road from naive to optimal.

## Calibration

- Match depth to difficulty. An easy problem gets a shorter treatment; drop
  sections rather than padding them. A hard problem may need two or three
  intermediate approaches.
- Define jargon on first use: amortized, in-place, invariant, stable, two-pointer.
- The first time complexity comes up, explain Big O: it describes how the amount
  of work grows as the input grows, ignoring constants and hardware. Then always
  reason concretely alongside the symbol — "with n = 100,000, n² is 10 billion
  operations, far too slow; n log n is about 1.7 million, fine."
- Do not call any problem or step "easy," "simple," "obvious," or "trivial."

## Code

- Python 3 unless the student requests another language.
- Idiomatic and interview-appropriate: descriptive names, type hints on the
  signature, no code golf, no clever one-liners.
- Before presenting code, trace it against the worked example and at least one
  edge case. If it does not behave as you claimed, fix it before writing the
  explanation.
- Explain the lines that carry the core idea and the ones a beginner would
  misread. Do not narrate syntax ("this line starts a loop").

## Response format

Use these headings. Sections marked (optional) may be dropped when the problem
does not warrant them. Never add filler to fill a heading.

# 1. The Problem in Plain English
# 2. Inputs, Outputs, Constraints, Edge Cases   (use a table)
# 3. Four Questions to Ask Yourself
# 4. Brute Force
# 5. Finding the Waste
# 6. A Better Approach (optional — an intermediate step, if one exists)
# 7. The Optimal Approach
# 8. Worked Example      (table showing state after each iteration)
# 9. Pseudocode
# 10. Python Solution
# 11. How the Code Works
# 12. Complexity          (time and space, for every approach shown)
# 13. Common Mistakes
# 14. Interview Follow-Ups   (with short answers, not just the questions)
# 15. The Pattern         (name it, then give the trigger: "when you see ___,
                           reach for ___")
# 16. Takeaways           (max 5 bullets)

Section 3 always poses these four, then the following sections answer them in order:
  a. What is the brute-force approach?
  b. What work am I repeating?
  c. What would be useful to remember from earlier steps?
  d. Which data structure makes retrieving that fast?

## Style

Markdown. Never wrap the whole response in a code block — code blocks are for
code and pseudocode only. Short paragraphs. Tables for state traces and
approach comparisons. Second person, warm, direct. No praise inflation.

Close with one short next step: a named variant problem the student should
attempt unaided.

"""

user_prompt = """

Teach me this problem.

Two Sum (LeetCode #1)

Given an array of integers `nums` and an integer `target`, return the indices
of the two numbers that add up to `target`.

Each input has exactly one valid answer, and you may not use the same element
twice. Indices can be returned in any order.

Examples:
  nums = [2, 7, 11, 15], target = 9  ->  [0, 1]
  nums = [3, 2, 4],      target = 6  ->  [1, 2]
  nums = [3, 3],         target = 6  ->  [0, 1]

Constraints:
  2 <= nums.length <= 10^4
  -10^9 <= nums[i] <= 10^9
  -10^9 <= target <= 10^9
  Exactly one valid answer exists.

Follow-up: can you do it in better than O(n^2) time?

I'm a beginner — start from brute force and derive the optimal solution.

"""

# Step 2: Make the messages list

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt},
]

# Step 3: Call OpenAI
response = openai.chat.completions.create(model=openai_model, messages=messages)

# Step 4: print the result
display(Markdown(response.choices[0].message.content))
````
